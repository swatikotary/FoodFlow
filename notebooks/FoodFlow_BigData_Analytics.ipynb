```python
import os

DATA_PATH = "/content/drive/MyDrive/BDA PROJECT"

print(os.listdir(DATA_PATH))
```

```python
orders_df = spark.read.csv(
    f"{DATA_PATH}/orders.csv",
    header=True,
    inferSchema=True
)

prior_df = spark.read.csv(
    f"{DATA_PATH}/order_products__prior.csv",
    header=True,
    inferSchema=True
)

train_df = spark.read.csv(
    f"{DATA_PATH}/order_products__train.csv",
    header=True,
    inferSchema=True
)

products_df = spark.read \
    .option("header", True) \
    .option("inferSchema", True) \
    .option("quote", '"') \
    .option("escape", '"') \
    .csv(f"{DATA_PATH}/products.csv")

aisles_df = spark.read.csv(
    f"{DATA_PATH}/aisles.csv",
    header=True,
    inferSchema=True
)

departments_df = spark.read.csv(
    f"{DATA_PATH}/departments.csv",
    header=True,
    inferSchema=True
)

print("All six datasets loaded successfully!")
```

```python
print("Orders:", orders_df.count())
print("Prior products:", prior_df.count())
print("Train products:", train_df.count())
print("Products:", products_df.count())
print("Aisles:", aisles_df.count())
print("Departments:", departments_df.count())
```

```python
# Join 1: Add customer and order details
prior_orders_df = prior_df.join(
    orders_df,
    on="order_id",
    how="inner"
)

# Join 2: Add product details
integrated_df = prior_orders_df.join(
    products_df,
    on="product_id",
    how="inner"
)

# Join 3: Add aisle names
integrated_df = integrated_df.join(
    aisles_df,
    on="aisle_id",
    how="inner"
)

# Join 4: Add department names
integrated_df = integrated_df.join(
    departments_df,
    on="department_id",
    how="inner"
)

print("Integrated dataset rebuilt successfully!")
print("Total rows:", integrated_df.count())
```

```python
integrated_df.printSchema()
integrated_df.show(5, truncate=False)
```

```python
print("Integrated rows:", integrated_df.count())
print("Total columns:", len(integrated_df.columns))

integrated_df.printSchema()
```

```python
from pyspark.sql.functions import count

basket_size_df = integrated_df.groupBy("order_id").agg(
    count("product_id").alias("basket_size")
)

basket_size_df.show(10)
```

```python
from pyspark.sql.functions import avg, round

average_basket_size = basket_size_df.agg(
    round(avg("basket_size"), 2).alias("average_basket_size")
)

average_basket_size.show()
```

```python
from pyspark.sql.functions import when, col

integrated_df = integrated_df.withColumn(
    "order_time_period",
    when(
        (col("order_hour_of_day") >= 5) &
        (col("order_hour_of_day") <= 11), "Morning"
    ).when(
        (col("order_hour_of_day") >= 12) &
        (col("order_hour_of_day") <= 16), "Afternoon"
    ).when(
        (col("order_hour_of_day") >= 17) &
        (col("order_hour_of_day") <= 20), "Evening"
    ).otherwise("Night")
)

integrated_df.select(
    "order_hour_of_day",
    "order_time_period"
).show(15)
```

```python
from pyspark.sql.functions import count, desc

time_period_summary = integrated_df.groupBy(
    "order_time_period"
).agg(
    count("order_id").alias("total_product_records")
).orderBy(
    desc("total_product_records")
)

time_period_summary.show()
```

```python
from pyspark.sql.functions import countDistinct, desc

customer_frequency_df = integrated_df.groupBy(
    "user_id"
).agg(
    countDistinct("order_id").alias("total_orders")
)

customer_frequency_df.show(10)
```

```python
from pyspark.sql.functions import when, col

integrated_df = integrated_df.withColumn(
    "is_first_order",
    when(col("order_number") == 1, 1).otherwise(0)
)

integrated_df.groupBy("is_first_order").count().show()
```

```python
from pyspark.sql.functions import avg, max, count

customer_orders_df = integrated_df.select(
    "user_id",
    "order_id",
    "order_number",
    "days_since_prior_order"
).dropDuplicates(["user_id", "order_id"])

customer_summary_df = customer_orders_df.groupBy(
    "user_id"
).agg(
    count("order_id").alias("total_orders"),
    round(avg("days_since_prior_order"), 2).alias("avg_days_between_orders"),
    max("order_number").alias("max_order_number")
)

customer_summary_df.show(10, truncate=False)
```

```python
from pyspark.sql.functions import desc

customer_summary_df.orderBy(
    desc("total_orders")
).show(15, truncate=False)
```

```python
from pyspark.sql.functions import sum, count, round, desc

product_features_df = integrated_df.groupBy(
    "product_id",
    "product_name"
).agg(
    count("*").alias("total_purchases"),
    sum("reordered").alias("reorder_count"),
    round(avg("reordered") * 100, 2).alias("reorder_percentage")
)

product_features_df.show(10, truncate=False)
```

```python
product_features_df.orderBy(
    desc("total_purchases")
).show(15, truncate=False)
```

```python
product_features_df.filter(
    col("total_purchases") >= 100
).orderBy(
    desc("reorder_percentage")
).show(15, truncate=False)
```

```python
aisle_features_df = integrated_df.groupBy(
    "aisle_id",
    "aisle"
).agg(
    count("*").alias("total_purchases"),
    countDistinct("product_id").alias("unique_products"),
    round(avg("reordered") * 100, 2).alias("reorder_percentage")
)

aisle_features_df.orderBy(
    desc("total_purchases")
).show(15, truncate=False)
```

```python
department_features_df = integrated_df.groupBy(
    "department_id",
    "department"
).agg(
    count("*").alias("total_purchases"),
    countDistinct("product_id").alias("unique_products"),
    round(avg("reordered") * 100, 2).alias("reorder_percentage")
)

department_features_df.orderBy(
    desc("total_purchases")
).show(25, truncate=False)
```

```python

# ==========================================
# FOODFLOW - CANDIDATE COVERAGE ANALYSIS
# ==========================================

from pyspark.sql.functions import col, countDistinct

# 1. Get historical orders
prior_orders = orders_df.filter(
    col("eval_set") == "prior"
).select("order_id", "user_id")

# 2. Create unique customer-product pairs
# These are products each customer purchased historically.
historical_candidates = (
    prior_df
    .join(prior_orders, on="order_id", how="inner")
    .select("user_id", "product_id")
    .distinct()
)

print("Historical customer-product candidates:")
print(historical_candidates.count())

# 3. Get actual products in the next train orders
train_order_users = train_orders.select("order_id", "user_id")

actual_next_products = (
    train_df
    .join(train_order_users, on="order_id", how="inner")
    .select("user_id", "product_id")
    .distinct()
)

print("Actual next-order customer-product pairs:")
print(actual_next_products.count())

# 4. Find products present in both datasets
matched_products = actual_next_products.join(
    historical_candidates,
    on=["user_id", "product_id"],
    how="inner"
)

# 5. Find products never purchased previously by that customer
new_products = actual_next_products.join(
    historical_candidates,
    on=["user_id", "product_id"],
    how="left_anti"
)

# 6. Calculate coverage
total_actual = actual_next_products.count()
matched_count = matched_products.count()
new_count = new_products.count()

print("\n========== CANDIDATE COVERAGE RESULTS ==========")
print("Total actual next-order products:", total_actual)
print("Previously purchased products:", matched_count)
print("New products:", new_count)

if total_actual > 0:
    print(
        "Historical candidate coverage:",
        round(matched_count / total_actual * 100, 2),
        "%"
    )

    print(
        "New product percentage:",
        round(new_count / total_actual * 100, 2),
        "%"
    )

print("\nSample new products:")
new_products.show(10, truncate=False)
```

```python
from pyspark.ml import Pipeline
from pyspark.ml.feature import (
    StringIndexer,
    OneHotEncoder,
    VectorAssembler,
    StandardScaler
)
from pyspark.sql.functions import log1p

# 1. Transform skewed numerical features
train_prepared = train_dataset.withColumn(
    "log_user_product_purchase_count",
    log1p(col("user_product_purchase_count"))
).withColumn(
    "log_product_total_purchases",
    log1p(col("product_total_purchases"))
)

validation_prepared = validation_dataset.withColumn(
    "log_user_product_purchase_count",
    log1p(col("user_product_purchase_count"))
).withColumn(
    "log_product_total_purchases",
    log1p(col("product_total_purchases"))
)

# 2. Define categorical encoders
aisle_indexer = StringIndexer(
    inputCol="aisle_id",
    outputCol="aisle_index",
    handleInvalid="keep"
)

department_indexer = StringIndexer(
    inputCol="department_id",
    outputCol="department_index",
    handleInvalid="keep"
)

onehot_encoder = OneHotEncoder(
    inputCols=["aisle_index", "department_index"],
    outputCols=["aisle_encoded", "department_encoded"],
    handleInvalid="keep"
)

# 3. Define numerical features
numeric_columns = [
    "log_user_product_purchase_count",
    "user_total_previous_orders",
    "user_product_reorder_ratio",
    "log_product_total_purchases",
    "user_product_recency",
    "has_purchased_before",
    "user_overall_reorder_rate"
]

# 4. Combine all features
assembler = VectorAssembler(
    inputCols=numeric_columns + [
        "aisle_encoded",
        "department_encoded"
    ],
    outputCol="raw_features",
    handleInvalid="error"
)

# 5. Scale assembled features
scaler = StandardScaler(
    inputCol="raw_features",
    outputCol="features",
    withStd=True,
    withMean=False
)

# 6. Build preprocessing pipeline
preprocessing_pipeline = Pipeline(stages=[
    aisle_indexer,
    department_indexer,
    onehot_encoder,
    assembler,
    scaler
])

# Fit ONLY on training data
preprocessing_model = preprocessing_pipeline.fit(train_prepared)

# Transform both datasets
train_ml = preprocessing_model.transform(train_prepared)
validation_ml = preprocessing_model.transform(validation_prepared)

print("Training data prepared:")
train_ml.select("features", "label", "class_weight").show(5)

print("Validation data prepared:")
validation_ml.select("features", "label").show(5)
```

```python
from pyspark.ml.evaluation import BinaryClassificationEvaluator

# 1. Generate predictions on validation data
predictions = lr_model.transform(validation_ml)

# 2. Calculate confusion matrix
confusion_matrix = (
    predictions
    .groupBy("label", "prediction")
    .count()
    .orderBy("label", "prediction")
)

confusion_matrix.show()

# 3. Calculate Precision and Recall
tp = predictions.filter(
    (col("label") == 1) & (col("prediction") == 1)
).count()

fp = predictions.filter(
    (col("label") == 0) & (col("prediction") == 1)
).count()

fn = predictions.filter(
    (col("label") == 1) & (col("prediction") == 0)
).count()

precision = tp / (tp + fp) if (tp + fp) > 0 else 0
recall = tp / (tp + fn) if (tp + fn) > 0 else 0
f1 = (
    2 * precision * recall / (precision + recall)
    if (precision + recall) > 0 else 0
)

# 4. Calculate PR-AUC
pr_evaluator = BinaryClassificationEvaluator(
    labelCol="label",
    rawPredictionCol="rawPrediction",
    metricName="areaUnderPR"
)

pr_auc = pr_evaluator.evaluate(predictions)

print("True Positives:", tp)
print("False Positives:", fp)
print("False Negatives:", fn)
print("-----------------------------")
print("Precision:", round(precision, 4))
print("Recall:", round(recall, 4))
print("F1 Score:", round(f1, 4))
print("PR-AUC:", round(pr_auc, 4))
```

```python
from pyspark.ml.functions import vector_to_array
from pyspark.sql.functions import sum as spark_sum

# Extract probability of class 1
threshold_data = predictions.withColumn(
    "positive_probability",
    vector_to_array("probability")[1]
)

thresholds = [0.3, 0.4, 0.5, 0.6, 0.7, 0.8]

# Build aggregation expressions for all thresholds
expressions = []

for t in thresholds:
    expressions.extend([
        spark_sum(
            when(
                (col("label") == 1) &
                (col("positive_probability") >= t), 1
            ).otherwise(0)
        ).alias(f"tp_{t}"),

        spark_sum(
            when(
                (col("label") == 0) &
                (col("positive_probability") >= t), 1
            ).otherwise(0)
        ).alias(f"fp_{t}")
    ])

# One aggregation over validation predictions
threshold_counts = threshold_data.agg(*expressions).collect()[0]

# Calculate metrics locally in Python
for t in thresholds:
    tp = threshold_counts[f"tp_{t}"]
    fp = threshold_counts[f"fp_{t}"]
    fn = 190209 - tp  # Total actual positives in validation

    precision = tp / (tp + fp) if tp + fp else 0
    recall = tp / (tp + fn) if tp + fn else 0
    f1 = (
        2 * precision * recall / (precision + recall)
        if precision + recall else 0
    )

    print(
        f"Threshold: {t} | "
        f"Precision: {precision:.4f} | "
        f"Recall: {recall:.4f} | "
        f"F1: {f1:.4f}"
    )
```

```python
from pyspark.ml.functions import vector_to_array
from pyspark.sql.window import Window

recommendation_scores = recommendation_predictions.withColumn(
    "reorder_probability",
    vector_to_array("probability")[1]
)

customer_window = Window.partitionBy("user_id").orderBy(
    col("reorder_probability").desc()
)

top_recommendations = (
    recommendation_scores
    .withColumn("rank", row_number().over(customer_window))
    .filter(col("rank") <= 10)
    .select(
        "user_id",
        "product_id",
        "reorder_probability",
        "rank"
    )
)

print("Top 10 recommendations generated!")
```

```python
top_recommendations_named = (
    top_recommendations
    .join(
        products_df.select("product_id", "product_name"),
        on="product_id",
        how="left"
    )
    .select(
        "user_id",
        "rank",
        "product_id",
        "product_name",
        "reorder_probability"
    )
)

top_recommendations_named.orderBy(
    "user_id", "rank"
).show(30, truncate=False)
```

```python
from pyspark.sql.functions import sum as spark_sum

# Count correct recommendations for each customer
recommendation_hits = (
    top_recommendations
    .join(
        candidate_features.select("user_id", "product_id", "label"),
        on=["user_id", "product_id"],
        how="inner"
    )
    .groupBy("user_id")
    .agg(
        spark_sum("label").alias("hits_at_10")
    )
)

# Count all actual products in each validation customer's next order
actual_products_count = (
    actual_next_products
    .join(validation_users, on="user_id", how="inner")
    .groupBy("user_id")
    .agg(
        count("*").alias("actual_product_count")
    )
)

# Combine both measurements for every validation customer
user_metrics = (
    validation_users
    .join(recommendation_hits, on="user_id", how="left")
    .join(actual_products_count, on="user_id", how="left")
    .fillna(0, subset=["hits_at_10", "actual_product_count"])
    .withColumn(
        "precision_at_10",
        col("hits_at_10") / 10.0
    )
    .withColumn(
        "recall_at_10",
        when(
            col("actual_product_count") > 0,
            col("hits_at_10") / col("actual_product_count")
        ).otherwise(0.0)
    )
)

# Calculate overall recommendation metrics
metrics = user_metrics.agg(
    count("*").alias("total_customers"),
    spark_sum("hits_at_10").alias("total_correct_recommendations"),
    spark_sum("actual_product_count").alias("total_actual_products"),
    avg("precision_at_10").alias("mean_precision_at_10"),
    avg("recall_at_10").alias("mean_recall_at_10"),
    (
        spark_sum("hits_at_10") /
        spark_sum("actual_product_count")
    ).alias("overall_recall_at_10")
).collect()[0]

print("Total validation customers:", metrics["total_customers"])
print("Correct recommendations:", metrics["total_correct_recommendations"])
print("Mean Precision@10:", round(metrics["mean_precision_at_10"] * 100, 2), "%")
print("Mean Recall@10:", round(metrics["mean_recall_at_10"] * 100, 2), "%")
print("Overall Recall@10:", round(metrics["overall_recall_at_10"] * 100, 2), "%")
```
